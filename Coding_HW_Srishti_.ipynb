```python
def max_val(input_list):
    """
    :param input_list: a nested list
    :return: the maximum value of each inner list
    """
    # Your code here
    return [max(inner) for inner in input_list]

L = [[1,2,5], [4,3,3], [1,7,6]]
result = max_val(L)
result
```

```python
def filter_list(input_list):
    """
    :param input_list: a nested list
    :return: all the lists whose length is 3 from a nested list
    """
    # Your code here
    return list(filter(lambda sub: len(sub) == 3, input_list))
L = [[2, 5, 4, 9], [2, 2], [3, 4, 5]]
filter_list(L)
```

```python
def expo_list(input_list):
    out = []
    if len(input_list) > 0:
        out.append(input_list[0])
    if len(input_list) > 1:
        out.append([x**2 for x in input_list[1]])
    if len(input_list) > 2:
        out.append([x**3 for x in input_list[2]])
    return out

L = [[1, 2], [2, 3, 4], [5, 6, 7]]
expo_list(L)
```

```python
def list_comp(input_list):
    """
    :param input_list: a list of integers
    :return: Assign `a` to the numbers greater than 8; `b` to the numbers less
    or equal to 8 and greater than 5; `c` to everything else
    """
    return ['a' if x > 8 else ('b' if x > 5 else 'c') for x in input_list]
print(list_comp(range(1, 11)))
```

```python
def register_student(first_name, last_name, *args, **kwargs):
    # middle name: first extra positional argument if present
    middle_name = args[0] if args else None

    semester = kwargs.get('semester')
    year = kwargs.get('year')
    course = kwargs.get('course')

    # build the name part
    if middle_name:
        name_part = f"{first_name} {middle_name} {last_name}"
    else:
        name_part = f"{first_name} {last_name}"

    # build the registration part
    parts = []
    if semester:
        parts.append(semester)
    if year:
        parts.append(str(year))
    if course:
        parts.append(course)

    if parts:
        return f"{name_part} is registered for {' '.join(parts)}"
    else:
        return f"{name_part} is registered"

# No middle name, with semester and year
register_student('Jane', 'Doe', semester='Summer', year='2023')

# With middle name (first extra positional arg), and all keywords
register_student('John', 'Doe', 'Jack', semester='Summer', year='2023', course='AC4RM')
```

```python
from datetime import datetime

def time_it(func):
    """
    A decorator that measures the execution time of `func` and
    returns a formatted string:
    "The function <name> takes <seconds> seconds"
    """
    def wrapper(*args, **kwargs):
        start = datetime.now()
        func(*args, **kwargs)
        end = datetime.now()
        elapsed = (end - start).total_seconds()
        return f"The function {func.__name__} takes {elapsed} seconds"
    return wrapper

@time_it
def my_function(n):
    total = 0
    for i in range(n):
        total += i

result = my_function(1000000)
print(result)
```

```python
import contextlib
from datetime import datetime

@contextlib.contextmanager
def time_it():
    start = datetime.now()
    try:
        # Run the code inside the with-block
        yield
    finally:
        end = datetime.now()
        elapsed = (end - start).total_seconds()
        print(f"The function takes {elapsed} seconds")
with time_it():
    total = 0
    for i in range(1_000_000):
        total += i
```
